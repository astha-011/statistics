###STATISTICS BASICS

1.. Explain the different types of data (qualitative and quantitative) and provide examples of each. Discuss nominal, ordinal, interval, and ratio scales.
->
1.Types of Data:**

A.Qualitative Data (Categorical)
Qualitative data describes **qualities or characteristics**. It cannot be measured with numbers but can be categorized based on traits.

Examples:
- Hair color (black, brown, blonde)
- Gender (male, female, non-binary)
- Types of cuisine (Italian, Chinese, Mexican)

B. Quantitative Data (Numerical)**
Quantitative data involves **numbers and measurable values**. It tells us "how much," "how many," or "how often."

Examples:
- Age (25 years)
- Height (5’10”)
- Number of pets (2 dogs)

2.Levels of Measurement (Scales):

These describe how data values relate to each other:
1.Nominal Scale (Qualitative)**
- **Definition:** Data are labels or names with **no meaningful order**.
- **Math operations:** Cannot be added or ordered.
- **Example:**
  - Marital status: Single, Married, Divorced
  - Types of cars: Toyota, Ford, BMW

2.Ordinal Scale (Qualitative or Quantitative)**
- **Definition:** Data have a meaningful **order**, but **differences between values aren’t measurable**.
- **Math operations:** You can rank, but can’t calculate exact differences.
- **Example:**
  - Movie ratings: Poor, Fair, Good, Excellent
  - Education level: High school, Bachelor’s, Master’s, Ph

3.Interval Scale (Quantitative)**
- **Definition:** Ordered data with **equal intervals** between values, but **no true zero**.
- **Math operations:** You can add/subtract, but ratios are meaningless.
- **Example:**
  - Temperature in Celsius or Fahrenheit (0°C doesn’t mean “no temperature”)
  - Dates on a calendar (year 0 doesn’t mean “no time”)

4.Ratio Scale (Quantitative)**
- **Definition:** Like interval data, but has a **true zero**, allowing for meaningful **ratios**.
- **Math operations:** All arithmetic operations apply.
- **Example:**
  - Weight (0 kg means no weight)
  - Age (a 20-year-old is twice as old as a 10-year-old)
  - Income ($0 means no income)

2.What are the measures of central tendency, and when should you use each? Discuss the mean, median, and mode with examples and situations where each is appropriate.
->1.Mean (Average)**
Definition:**
The mean is the **sum of all values divided by the number of values**.

**Formula:**
\[
\text{Mean} = \frac{\text{Sum of values}}{\text{Number of values}}
\]

Example:
Test scores: 70, 80, 90  
Mean = (70 + 80 + 90) / 3 = **80**

2.Median

Definition:**
The median is the **middle value** when data are ordered from smallest to largest.

- If odd number of values → middle one
- If even number of values → average of two middle values
Example:**
Data: 10, 15, 50  
Median = **15**

Data: 10, 15, 20, 50  
Median = (15 + 20)/2 = **17.5**

3.Mode**
Definition:**
The mode is the value that occurs **most frequently** in a data set.
Example:**
Data: 2, 3, 3, 4, 5  
Mode = **3**

Data: 1, 2, 3, 4, 5 (no repeated values)  
Mode = **No mode**

Data: 2, 2, 3, 3, 4  
Modes = **2 and 3 (bimodal)**

6.Explain the concept of skewness and its types. How does skewness affect the interpretation of data?
->Skewness describes the asymmetry or lopsidedness of a data distribution. In a perfectly symmetrical distribution (like a bell curve), the left and right sides mirror each other. But in real life, data often lean to one side — that’s where skewness comes in.
Types of Skewness:
➤ 1. Symmetrical Distribution (No Skew)
The mean, median, and mode are all equal.

The data is balanced on both sides of the center.

Example: Heights of adults in a large population.

➤ 2. Positive Skew (Right-Skewed)
The tail is longer on the right (higher values).

Mean > Median > Mode

Most values are low, but a few high ones pull the mean to the right.

Example: Income (a few people earn way more than most)

➤ 3. Negative Skew (Left-Skewed)
The tail is longer on the left (lower values).

Mean < Median < Mode

Most values are high, but a few low ones pull the mean to the left.

Example: Age at retirement (most retire around 60–65, but a few retire much earlier)

11.Explain what a random variable is and differentiate between discrete and continuous random variables.
->A random variable is a numerical outcome of a random phenomenon or experiment. In simpler terms, it is a variable whose value is determined by the outcome of a random event or process.
1.Discrete Random Variables
Definition: A discrete random variable can take on a finite or countably infinite number of distinct values. These values are usually integers or whole numbers.
2.Continuous Random Variables
Definition: A continuous random variable can take on an infinite number of possible values within a given range or interval. These values are not restricted to integers and can be any real number.

12.Provide an example dataset, calculate both covariance and correlation, and interpret the results.
->Certainly! Let’s work through an example where we calculate both **covariance** and **correlation** and interpret the results.

### Example Dataset:

We have a dataset with the **X** and **Y** values representing the number of hours studied and the exam scores for 5 students:

| Student | Hours Studied (X) | Exam Score (Y) |
|---------|-------------------|----------------|
| 1       | 2                 | 65             |
| 2       | 3                 | 70             |
| 3       | 5                 | 85             |
| 4       | 7                 | 95             |
| 5       | 8                 | 100            |

We want to calculate the **covariance** and **correlation** to understand the relationship between the number of hours studied and the exam score.

Step 1: Calculate Covariance**

Covariance measures the extent to which two variables change together. If the variables tend to increase together, the covariance will be positive. If one increases while the other decreases, the covariance will be negative. If there's no consistent pattern, the covariance will be close to 0.

The formula for covariance is:

\[
\text{Cov}(X, Y) = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{n}
\]

Where:
- \( X_i \) and \( Y_i \) are the individual values
- \( \bar{X} \) and \( \bar{Y} \) are the means of the X and Y values
- \( n \) is the number of data points

Step 1.1: Calculate the Means**

First, we calculate the **mean** of X and Y:

- \( \bar{X} = \frac{2 + 3 + 5 + 7 + 8}{5} = 5 \)
- \( \bar{Y} = \frac{65 + 70 + 85 + 95 + 100}{5} = 83 \)

Step 1.2: Calculate Covariance**

Now, we apply the covariance formula:

\[
\text{Cov}(X, Y) = \frac{(2-5)(65-83) + (3-5)(70-83) + (5-5)(85-83) + (7-5)(95-83) + (8-5)(100-83)}{5}
\]

Breaking it down:

\[
= \frac{(-3)(-18) + (-2)(-13) + (0)(2) + (2)(12) + (3)(17)}{5}
\]
\[
= \frac{54 + 26 + 0 + 24 + 51}{5}
\]
\[
= \frac{155}{5} = 31
\]

So, the **covariance** between hours studied and exam score is **31**.

---

### **Step 2: Calculate Correlation**

Correlation standardizes the covariance by dividing it by the product of the standard deviations of X and Y. This allows correlation to range between **-1** and **1**.

The formula for the **correlation coefficient** (\( r \)) is:

\[
r = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y}
\]

Where:
- \( \sigma_X \) and \( \sigma_Y \) are the standard deviations of X and Y

#### **Step 2.1: Calculate the Standard Deviations of X and Y**

The formula for standard deviation is:

\[
\sigma_X = \sqrt{\frac{\sum (X_i - \bar{X})^2}{n}}
\]

For **X** (hours studied):

\[
\sigma_X = \sqrt{\frac{(2-5)^2 + (3-5)^2 + (5-5)^2 + (7-5)^2 + (8-5)^2}{5}}
\]
\[
= \sqrt{\frac{9 + 4 + 0 + 4 + 9}{5}} = \sqrt{\frac{26}{5}} = \sqrt{5.2} \approx 2.28
\]

For **Y** (exam scores):

\[
\sigma_Y = \sqrt{\frac{(65-83)^2 + (70-83)^2 + (85-83)^2 + (95-83)^2 + (100-83)^2}{5}}
\]
\[
= \sqrt{\frac{324 + 169 + 4 + 144 + 289}{5}} = \sqrt{\frac{930}{5}} = \sqrt{186} \approx 13.65
\]

#### **Step 2.2: Calculate Correlation**

Now, we can calculate the correlation:

\[
r = \frac{31}{(2.28)(13.65)} = \frac{31}{31.11} \approx 0.997
\]

So, the **correlation coefficient** between hours studied and exam score is **0.997**.
